# JAX numpy basics

```python
import jax.numpy as jnp
```

## Array Initialization

```python
jnp.array([1, 2, 3])
```

```python
jnp.zeros((3, 4)), jnp.ones((3, 4)), jnp.full((3, 4), 5), jnp.empty((3, 4))
```

```python
jnp.arange(10), jnp.linspace(0, 1, 10), jnp.eye(3)
```

## Shape Properties and Operations

```python
i = jnp.array([[1, 2, 3], [4, 5, 6]])
print(i.shape)
print(i.reshape((1, 6)))
print(i.flatten())
print(jnp.transpose(i))
print(i.T)
print(jnp.squeeze(i))
print(jnp.expand_dims(i, axis=0))
```

## Array Operations

```python
x = jnp.array([1.0, 2.0, 3.0])
y = jnp.array([4.0, 5.0, 6.0])

print(x + y)           # add
print(x * y)           # multiply
print(x ** 2)          # power
print(jnp.sqrt(x))     # sqrt
print(jnp.abs(-x))     # abs
print(jnp.exp(x))      # exp
print(jnp.log(x))      # log
print(jnp.dot(x, y))   # dot
A = jnp.array([[1.0, 2.0], [3.0, 4.0]])
B = jnp.array([[5.0, 6.0], [7.0, 8.0]])
print(jnp.matmul(A, B))  # matmul
```

## Aggregation Operations

```python
j = jnp.array([[1, 2, 3], [4, 5, 6]])

print(jnp.sum(j))
print(jnp.mean(j))
print(jnp.max(j))
print(jnp.min(j))
print(jnp.std(j))
print(jnp.var(j))
print(jnp.argmax(j))
print(jnp.argmin(j))
print(jnp.cumsum(j))
```

## Indexing and Slicing

```python
k = jnp.arange(12).reshape((3, 4))
print(k)
print(k[0])        # first row
print(k[:, 1])     # second column
print(k[1:, 2:])   # submatrix
print(k[0, -1])    # last element of first row
```

## Trigonometric Functions

```python
angles = jnp.linspace(0, jnp.pi, 5)
print(jnp.sin(angles))
print(jnp.cos(angles))
print(jnp.tan(angles))
```

## Linear Algebra

```python
M = jnp.array([[1.0, 2.0], [3.0, 4.0]])

print(jnp.linalg.inv(M))       # inv
print(jnp.linalg.det(M))       # det
print(jnp.linalg.norm(M))      # norm
print(jnp.linalg.eig(M))       # eig (eigenvalues, eigenvectors)
U, S, Vt = jnp.linalg.svd(M)
print(U, S, Vt)                 # svd
```
